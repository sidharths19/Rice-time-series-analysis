Rice accounts for approximately 46% of total food grain production in India and
serves as the staple food for nearly 60% of the country’s population. 
Rice is predominantly cultivated during the Kharif season, making it highly
dependent on monsoon rainfall and seasonal climatic conditions.

The dataset consists of rice production data for the period from 1966 to 2025.
The data is seasonal in nature and includes observations for both Kharif and

## Data Source

The data has been obtained from the UPAg (Unified Portal for Agricultural
Statistics) portal. This portal provides agricultural statistics in spreadsheet
format and includes detailed information on area, production, and yield
(productivity) for various crops across India.

The main objectives of this study are:
1. To examine how rice production has changed over time in India.
2. To analyze seasonal patterns and long-term trends in rice production.
3. To prepare the data for time series modeling.
4. To apply forecasting models for predicting future rice production and
   compare their performance using appropriate evaluation metrics.

## Importing data

```python
import plotly.express as px
import pandas as pd
from statsmodels.graphics.tsaplots import plot_pacf
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("All-India-profile_-Crop-wise-Productionfinal.csv")
```

## data cleaning

The main drawback of the data obtained from the UPAg portal is that it is not
directly suitable for time series analysis. The data requires several
preprocessing steps, including:

1. Reshaping the data into a time series format.
2. Handling seasonal observations (Kharif and Rabi).
3. Cleaning column names and removing inconsistencies.
4. Creating a datetime variable using year, month, and day.
5. Setting the datetime variable as the index.


```python
long_df = data.melt(
    id_vars=["Crop", "Season"],
    var_name="Year",
    value_name="Production"
)
```

```python
long_df["Year"] = long_df["Year"].str.extract(r'(\d{4})').astype(int)
```

```python
season_month = {
    "Kharif": 6,
    "Rabi": 11
}

long_df["Month"] = long_df["Season"].map(season_month)
```

## Creating Time stamp

```python
long_df["Date"] = pd.to_datetime(
    dict(year=long_df["Year"],
         month=long_df["Month"],
         day=1)
)
```

```python
ts_df = long_df.sort_values("Date").set_index("Date")
```

```python
Rice = ts_df[[ "Production"]]
print(Rice.head())
```
